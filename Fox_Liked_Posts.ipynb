```python
for item in data[0]['label_values']:
    if 'label' in item:
        row[item['label']] = item['value']
    elif item.get('title') == 'Owner':
        owner_fields = item['dict'][0]['dict']
        for owner_item in owner_fields:
            row['Owner ' + owner_item['label']] = owner_item['value']
row
```

```python
tidy_rows = []

for entry in data: 
    row = {}
    row['timestamp'] = entry['timestamp']
    
    for item in entry['label_values']:
        if 'label' in item:
            row[item['label']] = item['value']
        elif item.get('title') == 'Owner':
            owner_fields = item['dict'][0]['dict']
            for owner_item in owner_fields:
                row['Owner ' + owner_item['label']] = owner_item['value']

    tidy_rows.append(row)
tidy_rows[2]
```

```python
likes_df = pd.DataFrame(tidy_rows)
likes_df.head(10)
```

## Hypothesis

### "I comment on Instagram posts at a higher daily rate on weekends than on weekdays."

### Theoretical element
I am testing this theory because my school and work life greatly impact my social media usage, so I want to test the data on how much I comment on posts, especially on weekends

### Statistical element
I can use the same Date and time values to determine when I made the comments on Instagram posts

```python
import json
import pandas as pd

with open("post_comments_1.json", "r", encoding="utf-8") as f:
    comments_data = json.load(f)

print("Type:", type(comments_data))
print("Number of comments:", len(comments_data))

comments_data[0]
```

```python
def fix_text(text):
    try:
        return text.encode("latin-1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return text
#I looked up how to turn the codes into the emojis that I used in my comments


rows = []

for record in comments_data:
    info = record["string_map_data"]
#I kept having errors thrown, so Claude helped me understand that this data is in a different format than the others and figure out how I should format it

    comment_info = info.get("Comment", {})
    owner_info = info.get("Media Owner", {})
    time_info = info.get("Time", {})

    rows.append({
        "comment": fix_text(comment_info.get("value", "")),
        "media_owner": owner_info.get("value", ""),
        "timestamp": time_info.get("timestamp")
    })

comments_df = pd.DataFrame(rows)

comments_df["DateTime"] = pd.to_datetime(comments_df["timestamp"], unit="s")

print("Shape:", comments_df.shape)
comments_df.head(10)
```
